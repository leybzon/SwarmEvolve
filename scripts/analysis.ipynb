# SwarmEvolve analysis notebook (M12)

Thin wrapper around `scripts/analysis.py` — the real logic lives in
that module so it is unit-testable without a Jupyter kernel. Use this
notebook to explore a completed tournament or evolution run
interactively; the single code cell below runs the whole analysis
pipeline and displays the resulting plots inline.

## Usage

Edit `TOURNAMENT_DIR` / `EVOLUTION_DIR` / `OUT_DIR` to point at your
experiment directories, then *Restart & Run All*.


```python
from pathlib import Path
import sys

REPO_ROOT = Path().resolve().parent
sys.path.insert(0, str(REPO_ROOT / "scripts"))

import analysis

# --- EDIT ME -------------------------------------------------------
TOURNAMENT_DIR = REPO_ROOT / "data" / "tournaments" / "rr-4ai"
EVOLUTION_DIR = None  # e.g. REPO_ROOT / "data" / "evolve" / "run001"
OUT_DIR = REPO_ROOT / "data" / "analysis" / "notebook"
# -------------------------------------------------------------------

OUT_DIR.mkdir(parents=True, exist_ok=True)
if TOURNAMENT_DIR and TOURNAMENT_DIR.is_dir():
    analysis.analyse_tournament(TOURNAMENT_DIR, OUT_DIR / "tournament")
if EVOLUTION_DIR and EVOLUTION_DIR.is_dir():
    analysis.analyse_evolution(EVOLUTION_DIR, OUT_DIR / "evolution")

from IPython.display import Image, display
for p in sorted((OUT_DIR / "tournament").glob("*.png")):
    display(Image(str(p)))
```
